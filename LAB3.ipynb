```python
df.replace({'y':{'no':0,'yes':1}}, inplace=True)
```

```python
for column in ['job', 'education', 'contact', 'poutcome']:
    if column in df.columns:
        df[column] = df[column].fillna('unknown')
df.isnull().sum()
```

```python
# Drop the 'duration' column due to data leakage
df = df.drop('duration', axis=1)

# Display the columns to confirm 'duration' is removed
print(df.columns)
```

```python

categorical_cols = ['job', 'marital', 'education', 'contact', 'month', 'poutcome', 'default', 'housing', 'loan']

df = pd.get_dummies(df, columns=categorical_cols, drop_first=True)

# Display the first few rows of the updated DataFrame and its info to see the new columns
display(df.head())
df.info()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

# Separate features (X) and target (y) from the preprocessed DataFrame
X = df.drop('y', axis=1)
y = df['y']

# Split the data into training and testing sets using the preprocessed X
x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=100)

# Initialize and fit the Logistic Regression model
lg_model = LogisticRegression(max_iter=1000) # Increased max_iter for convergence
lg_model.fit(x_train, y_train)

print("Logistic Regression model fitted successfully.")
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

# Make predictions on the test set
y_pred = lg_model.predict(x_test)

# Calculate evaluation metrics
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print(f"Accuracy: {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1-Score: {f1:.4f}")

# Display the confusion matrix
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
def sigmoid(z):
    z=np.array(z)
    return 1/(1+np.exp(-z))

def train_logistics_regression(X,y,lr=0.01,Epochs=1000):
    x=np.array(X,dtype=np.float64)
    y=np.array(y,dtype=np.float64).reshape(-1, 1) # Reshape y to be a column vector
    m,n=X.shape
    w=np.zeros((n,1))
    for epoch_idx in range(Epochs):
        z=np.dot(x,w)
        pred=sigmoid(z)
        gradient=(1/m)*(np.dot(x.T,(pred-y)))
        w=w-lr*gradient
        if epoch_idx % 1000==0:
            loss=-(1/m)*(np.dot(y.T,np.log(pred+1e-9))+np.dot((1-y).T,np.log(1-pred+1e-9)))
            print(f"Epoch:{epoch_idx} loss:{loss.item():.4f}")
    return w
```

```python
def predict(X,w,threshold=0.5):
  X=np.array(X,dtype=np.float64)
  return (sigmoid(np.dot(X,w))>=threshold).astype(int)
```

```python
print(f"Shape of x_train: {x_train.shape}")
print(f"Shape of y_train: {y_train.shape}")
weights=train_logistics_regression(x_train,y_train)
print(f"Shape of weights: {weights.shape}")

print(f"Shape of x_test: {x_test.shape}")
print(f"Shape of y_test: {y_test.shape}")
y_pred_raw = predict(x_test,weights)
print(f"Shape of y_pred before flatten: {y_pred_raw.shape}")
y_pred=y_pred_raw.flatten() # Flatten y_pred to match y_test shape
print(f"Shape of y_pred after flatten: {y_pred.shape}")

accuracy=accuracy_score(y_test,y_pred)
print(f"Accuracy:{accuracy:.4f}")
```
